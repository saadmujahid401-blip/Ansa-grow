</span>
<br>
<code style="color:#38bdf8; font-weight:bold;">Rescaling</code> layers apply ki hain taake pixel values [0, 255] se [0, 1] range mein normalize ho sakein, sath hi <code style="color:#38bdf8; font-weight:bold;">cache()</code> aur <code style="color:#38bdf8; font-weight:bold;">prefetch()</code> ki madad se I/O bottleneck ko khatam karke training speed ko optimize kiya hai.
</div>

</div>

```python
# Configure dataset performance optimization parameters
AUTOTUNE = tf.data.AUTOTUNE  # Automatically tune buffer sizes for optimal performance

# Cache training dataset in memory/disk to avoid re-reading files during epochs
train_ds = train_ds.cache().prefetch(buffer_size=AUTOTUNE)

# Cache validation dataset for faster evaluation processing
val_ds = val_ds.cache().prefetch(buffer_size=AUTOTUNE)

# Define a normalization layer to scale pixel values from [0, 255] to [0.0, 1.0]
normalization_layer = tf.keras.layers.Rescaling(1./255)

# Apply normalization layer to training images
train_ds = train_ds.map(lambda x, y: (normalization_layer(x), y))

# Apply normalization layer to validation images
val_ds = val_ds.map(lambda x, y: (normalization_layer(x), y))

print("Step 2 Completed: Data normalization and prefetching pipeline optimized successfully!")
```

Multi-class classification ke liye ek robust <code style="color:#38bdf8; font-weight:bold;">Sequential CNN model</code> build kiya hai, jismein multiple Conv2D layers, MaxPooling layers, Dropout regularization aur final softmax activation layer shamil hain jo 5 animal classes ko predict karengi.
</div>

```python
# Initialize a Sequential model container for stacking neural network layers
model = tf.keras.Sequential([
    # First Convolutional Block with 32 filters and ReLU activation
    tf.keras.layers.Conv2D(32, 3, padding='same', activation='relu', input_shape=(180, 180, 3)),
    # Max pooling layer to downsample spatial dimensions
    tf.keras.layers.MaxPooling2D(),
    
    # Second Convolutional Block with 64 filters
    tf.keras.layers.Conv2D(64, 3, padding='same', activation='relu'),
    # Max pooling layer
    tf.keras.layers.MaxPooling2D(),
    
    # Third Convolutional Block with 128 filters to capture complex features
    tf.keras.layers.Conv2D(128, 3, padding='same', activation='relu'),
    # Max pooling layer
    tf.keras.layers.MaxPooling2D(),
    
    # Dropout layer to prevent overfitting by randomly dropping 30% of neurons
    tf.keras.layers.Dropout(0.3),
    # Flatten multi-dimensional feature maps into 1D vector
    tf.keras.layers.Flatten(),
    # Dense hidden layer with 128 units
    tf.keras.layers.Dense(128, activation='relu'),
    # Output layer with 5 units corresponding to the 5 animal species, using Softmax for multi-class probability
    tf.keras.layers.Dense(len(class_names), activation='softmax')
])

# Compile the model with Adam optimizer, Sparse Categorical Crossentropy loss, and Accuracy metric
model.compile(optimizer='adam', loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=False), metrics=['accuracy'])

# Display model architecture summary
model.summary()
```
