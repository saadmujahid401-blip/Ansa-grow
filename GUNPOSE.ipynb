<p style="margin:0; font-size:16px; line-height:1.8; color:#ffffff; font-weight:600;">
<span style="display:inline-block; padding:6px 16px; background:#1e293b; color:#ffffff; border-radius:20px; font-size:14px; font-weight:800; letter-spacing:0.5px; box-shadow:0 0 10px rgba(0,0,0,0.3); border:1.5px solid #eab308; margin-right:8px; vertical-align:middle;">📌 <span style="color:#38bdf8;">Adil ne is step 1 mein:</span></span>

```python
import os
import tensorflow as tf

# Define dataset directory path to access images
dataset_path = "Gun_Pose_Image_Classification_Dataset"

# Check if the dataset folder exists on the system
if os.path.exists(dataset_path):
    print(f"✅ Dataset Found: {dataset_path}")
    # List all class folders inside the main dataset directory
    print(f"📁 Classes: {os.listdir(dataset_path)}")
else:
    print(f"❌ Error: '{dataset_path}' folder not found!")

# Set standard image dimensions for CNN model compatibility
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
print("🚀 Step 01 Completed Successfully!")
```

⚡ STEP 02 • DATA AUGMENTATION & PREPROCESSING
</span>
<h2 style="margin:0; color:#000000; font-size:26px; font-weight:900; letter-spacing:0.5px;">
🎯 Image Preprocessing & Data Generators Pipeline
</h2>
</div>
<div style="margin:25px;">
<div style="background:#000000; border:3px solid #eab308; border-radius:18px; padding:20px 25px; box-shadow:0 4px 15px rgba(0,0,0,0.2); text-align:left;">
<p style="margin:0; font-size:16px; line-height:1.8; color:#ffffff; font-weight:600;">
<span style="display:inline-block; padding:6px 16px; background:#1e293b; color:#ffffff; border-radius:20px; font-size:14px; font-weight:800; letter-spacing:0.5px; box-shadow:0 0 10px rgba(0,0,0,0.3); border:1.5px solid #eab308; margin-right:8px; vertical-align:middle;">📌 <span style="color:#38bdf8;">Adil ne is step 2 mein:</span></span> 
ImageDataGenerator ka istemal karte hue training images par robust data augmentation (jaise rotation, zoom, aur horizontal flip) apply ki hai aur pixel values ko normalize karke train aur validation datasets successfully prepare kiye hain.
</p>
</div>
</div>
</div>

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. ImageDataGenerator setup with rescaling and augmentation to prevent overfitting
datagen = ImageDataGenerator(rescale=1./255, rotation_range=25, zoom_range=0.2, horizontal_flip=True, validation_split=0.2)

# 2. Create training generator pointing to our dataset directory
train_generator = datagen.flow_from_directory(dataset_path, target_size=(224, 224), batch_size=32, class_mode='binary', subset='training', shuffle=True)

# 3. Create validation generator for model testing
val_generator = datagen.flow_from_directory(dataset_path, target_size=(224, 224), batch_size=32, class_mode='binary', subset='validation', shuffle=False)

print("🚀 Step 02 Completed Successfully: Generators are ready!")
```

⚡ STEP 03 • MODEL BUILDING & TRANSFER LEARNING
</span>
<h2 style="margin:0; color:#000000; font-size:26px; font-weight:900; letter-spacing:0.5px;">
🎯 MobileNetV2 Architecture & Classification Head
</h2>
</div>
<div style="margin:25px;">
<div style="background:#000000; border:3px solid #eab308; border-radius:18px; padding:20px 25px; box-shadow:0 4px 15px rgba(0,0,0,0.2); text-align:left;">
<p style="margin:0; font-size:16px; line-height:1.8; color:#ffffff; font-weight:600;">
<span style="display:inline-block; padding:6px 16px; background:#1e293b; color:#ffffff; border-radius:20px; font-size:14px; font-weight:800; letter-spacing:0.5px; box-shadow:0 0 10px rgba(0,0,0,0.3); border:1.5px solid #eab308; margin-right:8px; vertical-align:middle;">📌 <span style="color:#38bdf8;">Adil ne is step 3 mein:</span></span> 
MobileNetV2 base model ko pre-trained ImageNet weights ke sath load kiya hai, uske base layers ko freeze karke custom classification head add kiya hai, aur binary classification ke liye model compile kiya hai.
</p>
</div>
</div>
</div>

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Sequential
from tensorflow.keras.optimizers import Adam

# 1. Load pre-trained MobileNetV2 base model without top classification layers and set image input shape
base_model = MobileNetV2(input_shape=(224, 224, 3), include_top=False, weights='imagenet')

# 2. Freeze the base model weights to retain learned feature extraction from ImageNet
base_model.trainable = False

# 3. Build a sequential model by combining the base model and custom dense classification layers
model = Sequential([base_model, GlobalAveragePooling2D(), Dense(1, activation='sigmoid')])

# 4. Compile the model specifying the optimizer, loss function, and evaluation metrics for binary classification
model.compile(optimizer=Adam(learning_rate=0.001), loss='binary_crossentropy', metrics=['accuracy'])

# 5. Display a summary of the model architecture to verify layer connections and parameters
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint

# 1. Setup EarlyStopping to halt training if validation loss stops improving, preventing overfitting
early_stop = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

# 2. Setup ModelCheckpoint to automatically save the model file with the highest validation accuracy
checkpoint = ModelCheckpoint('best_gun_model.keras', monitor='val_accuracy', save_best_only=True)

# 3. Train the model using the training generator and evaluate on validation data over 10 epochs
history = model.fit(
    train_generator,
    validation_data=val_generator,
    epochs=10,
    callbacks=[early_stop, checkpoint]
)

print("🚀 Step 04 Completed Successfully: Model training finished and best weights saved!")
```
