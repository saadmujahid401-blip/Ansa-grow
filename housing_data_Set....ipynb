```python
# Zaroori data manipulation aur mathematical operations ke liye pandas aur numpy import kiye hain
import pandas as pd
import numpy as np

# Data visualization (graphs aur plots) banane ke liye matplotlib aur seaborn import kiye hain
import matplotlib.pyplot as plt
import seaborn as sns

# California Housing CSV file ko read karke pandas DataFrame ke andar load kiya gaya hai
df = pd.read_csv('housing.csv')

# Dataset ki pehli 5 rows ko display kiya gaya hai taake houses ke features ka andaza ho sakay
display(df.head())
```

    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 2 • EDA & Missing Values</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Dataset Analysis & Missing Values Inspection</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein dataset ki structural info, data types, aur columns ke andar mojood missing values ko check kiya gaya hai taake safai ka amal shuru kiya ja sakay.
    </p>
</div>

```python
# Dataset ki general information (rows, columns, data types, aur non-null counts) check karne ke liye
df.info()

# Har column mein kitne missing (null) values hain, unka total sum maloom kiya gaya hai
print("\n--- Missing Values Count ---")
print(df.isnull().sum())

# Dataset ka statistical summary (mean, min, max, standard deviation) dekhne ke liye
display(df.describe())
```

<div style="background: linear-gradient(135deg, #090d16 0%, #111827 50%, #1e1b4b 100%); color: #ffffff; padding: 28px; border-radius: 16px; margin-bottom: 25px; box-shadow: 0 15px 35px rgba(0, 0, 0, 0.5); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; border: 2px solid #D4AF37; border-left: 8px solid #FFD700;">
    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 3 • Data Preprocessing</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Handling Missing Values & Feature Splitting</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein <code>total_bedrooms</code> ke missing values ko median se fill kiya gaya hai aur Features (X) ko Target (y) se alag kiya gaya hai.
    </p>
</div>

```python
# total_bedrooms column ke missing values ko median value ke sath replace kar rahe hain
median_bedrooms = df['total_bedrooms'].median()
df['total_bedrooms'].fillna(median_bedrooms, inplace=True)

# Dobara check kar rahe hain ke ab koi missing value baqi toh nahi bachi
print("Remaining Missing Values:", df['total_bedrooms'].isnull().sum())

# Features (X) aur Target variable (y - median_house_value) ko alag alag kar rahe hain
X = df.drop(columns=['median_house_value'])
y = df['median_house_value']

# Check karne ke liye ke X aur y ka shape thek hai ya nahi
print("Features Shape (X):", X.shape)
print("Target Shape (y):", y.shape)
```

<div style="background: linear-gradient(135deg, #090d16 0%, #111827 50%, #1e1b4b 100%); color: #ffffff; padding: 28px; border-radius: 16px; margin-bottom: 25px; box-shadow: 0 15px 35px rgba(0, 0, 0, 0.5); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; border: 2px solid #D4AF37; border-left: 8px solid #FFD700;">
    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 4 • Encoding & Scaling</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Categorical Encoding & Feature Scaling</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein categorical column (ocean_proximity) ko One-Hot Encoding ke zariye numbers mein convert kiya gaya hai aur data ko Standard Scaler se scale kiya gaya hai.
    </p>
</div>

```python
# Categorical column (ocean_proximity) ko One-Hot Encoding ke zariye numerical columns mein convert kar rahe hain
X = pd.get_dummies(X, columns=['ocean_proximity'], drop_first=True)

# Scikit-learn se StandardScaler import kar rahe hain feature scaling ke liye
from sklearn.preprocessing import StandardScaler

# StandardScaler ka object bana rahe hain
scaler = StandardScaler()

# Features (X) ke oopar scaling apply kar rahe hain taake saare features aik hi range mein aa jayein
X_scaled = scaler.fit_transform(X)

# Scaled data ka pehla row check karne ke liye
print("Scaled Features Example:")
print(X_scaled[0])
```

<div style="background: linear-gradient(135deg, #090d16 0%, #111827 50%, #1e1b4b 100%); color: #ffffff; padding: 28px; border-radius: 16px; margin-bottom: 25px; box-shadow: 0 15px 35px rgba(0, 0, 0, 0.5); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; border: 2px solid #D4AF37; border-left: 8px solid #FFD700;">
    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 5 • Train-Test Split</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Splitting Dataset into Training & Testing Sets</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein dataset ko 80% training data aur 20% testing data mein divide kiya gaya hai taake model ki generalization check ki ja sakay.
    </p>
</div>

```python
# Model training aur testing ke liye train_test_split function import kar rahe hain
from sklearn.model_selection import train_test_split

# Dataset ko 80% training aur 20% testing parts mein split kar rahe hain
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.20, random_state=42)

# Dono sets ki rows check karne ke liye
print("Training Data Shape (X_train):", X_train.shape)
print("Testing Data Shape (X_test):", X_test.shape)
```

<div style="background: linear-gradient(135deg, #090d16 0%, #111827 50%, #1e1b4b 100%); color: #ffffff; padding: 28px; border-radius: 16px; margin-bottom: 25px; box-shadow: 0 15px 35px rgba(0, 0, 0, 0.5); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; border: 2px solid #D4AF37; border-left: 8px solid #FFD700;">
    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 6 • Model Training</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Training Multiple Regression Algorithms</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein Linear Regression, Ridge Regression, Random Forest, aur Gradient Boosting models ko train kiya gaya hai.
    </p>
</div>

```python
# Zaroori regression models import kar rahe hain
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

# Har model ka aik object (instance) bana rahe hain
linear_model = LinearRegression()
ridge_model = Ridge(alpha=1.0)
rf_model = RandomForestRegressor(n_estimators=50, random_state=42)
gb_model = GradientBoostingRegressor(n_estimators=50, random_state=42)

# Models ko training data (X_train, y_train) par fit (train) kar rahe hain
linear_model.fit(X_train, y_train)
ridge_model.fit(X_train, y_train)
rf_model.fit(X_train, y_train)
gb_model.fit(X_train, y_train)

# Success confirmation message
print("✅ Step 6 Completed: All models have been successfully trained on training data!")
```

<div style="background: linear-gradient(135deg, #090d16 0%, #111827 50%, #1e1b4b 100%); color: #ffffff; padding: 28px; border-radius: 16px; margin-bottom: 25px; box-shadow: 0 15px 35px rgba(0, 0, 0, 0.5); font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; border: 2px solid #D4AF37; border-left: 8px solid #FFD700;">
    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 7 • Model Evaluation</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Calculating MAE, MSE, RMSE, & R² Score</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein trained models ki predictions nikal kar MAE, MSE, RMSE aur R² Score calculate kiye gaye hain taake performance measure ho sakay.
    </p>
</div>

```python
# Evaluation metrics ke liye functions import kar rahe hain
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Trained models ki dictionary bana rahe hain
models_dict = {
    "Linear Regression": linear_model,
    "Ridge Regression": ridge_model,
    "Random Forest": rf_model,
    "Gradient Boosting": gb_model
}

evaluation_results = []

for name, model in models_dict.items():
    # Test data par predictions nikal rahe hain
    y_pred = model.predict(X_test)
    
    # Metrics calculate kar rahe hain
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)
    
    # Results list mein store kar rahe hain
    evaluation_results.append({
        "Model Name": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R² Score": r2
    })

# Professional DataFrame bana kar display kar rahe hain
results_df = pd.DataFrame(evaluation_results)
display(results_df)
```

    <div style="text-align: center; border-bottom: 2px solid rgba(212, 175, 55, 0.3); padding-bottom: 14px; margin-bottom: 16px;">
        <span style="background-color: #FFD700; color: #090d16; padding: 5px 15px; border-radius: 20px; font-weight: 800; font-size: 12px; text-transform: uppercase; letter-spacing: 1.5px; box-shadow: 0 4px 12px rgba(255, 215, 0, 0.4);">Step 8 • Hyperparameter Tuning</span>
        <h2 style="margin: 12px 0 5px 0; font-size: 24px; color: #FFD700; font-weight: 800; letter-spacing: 0.5px;">Optimizing Random Forest using GridSearchCV</h2>
    </div>
    <p style="margin: 0; line-height: 1.6; color: #E5E7EB; font-size: 14px;">
        <strong style="color: #FFD700;">Code Summary:</strong> Is code mein GridSearchCV ka istemal karke Random Forest model ke liye best hyperparameters (jaise n_estimators aur max_depth) find kiye gaye hain.
    </p>
</div>

```python
# GridSearchCV import kar rahe hain hyperparameter tuning ke liye
from sklearn.model_selection import GridSearchCV

# Parameters ka grid set kar rahe hain jo test honge
param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [10, 20, None],
    'min_samples_split': [2, 5]
}

# Base model instantiate kar rahe hain
rf_base = RandomForestRegressor(random_state=42)

# GridSearchCV object bana rahe hain (CV=3 rakha hai taake fast run ho)
grid_search = GridSearchCV(
    estimator=rf_base, 
    param_grid=param_grid, 
    cv=3, 
    scoring='r2', 
    n_jobs=-1
)

print("⏳ Grid Search tuning shuru ho gayi hai, thoda waqt lag sakta hai...")

# Training data par grid search fit kar rahe hain
grid_search.fit(X_train, y_train)

# Best parameters aur best score print kar rahe hain
print("\n✅ Best Hyperparameters Found:")
print(grid_search.best_params_)

print("\n🏆 Best Cross-Validation R² Score:")
print(round(grid_search.best_score_, 4))

# Best model ko variable mein save kar rahe hain
best_rf_model = grid_search.best_estimator_
```
